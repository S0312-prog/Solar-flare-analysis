# **Solar Low Energy X-ray Spectrometer (SoLEXS)**

## **Introduction**

Aditya-L1 mission was launched on $2^{nd}$ September, 2023 with an objective to study the dynamic nature of the Sun and the solar corona. It has total of $7$ payloads (4 remote sensing and 3 in-situ instruments) one of which is SoLEXS, which stands for **So**lar **L**ow **E**nergy **X**-ray **S**pectrometer.

SoLEXS was designed to study the solar flares through spectroscopy and photometry in the soft X-ray regime. SoLEXS with its moderate spectral resolution and large energy range of $1-22$ $keV$ allows to study the solar abundances by appropriately accounting for the continuum. Further, SoLEXS along with VELC can be used to study the relation between flares and CMEs and along with SUIT to study flare-prominance relations which aid in understanding the energy release mechanism involved in coronal heating.

During solar flares, the photon flux observed can vary from $~20 - 2\times10^6$ $photons/s/mm^2$. Hence, to study such events with high range in photon flux, detectors with high dynamic range needs to be used, Thus, for SoLEXS **Silicon Drift Detectors** (SDDs) are used, which provide
- good soectroscopic performance
- low anode capacitance
- low leakage current
- can be cooled using a simple Peltier cooler.

## **Data Products**

The inital data received from the instrument will be in binary format and is called the Level-0 (L0) data product. Necessary calibrations are applied on this L0 data to bring it into usable form for scientific analysis which is called Level-1 (L1) data. The two L1 data main products of SoLEXS are:
- Light Curve file ($.lc$ file)
- Type II Pulse Height Amplitude (PHA) file ($.pi$ file)

The Type II PHA file essentially contains spectrum at equally spaced intervals and it facilitates selecting spectra corresponding to desired time periods for integrating to get a spectrum with improved signal-to-noise ratio (SNR).

In this tutorial hands-on session, you'll be working with the L1 data products of SoLEXS corresponding to $12^{th}$ February, 2024.

## **Spectral Fitting and Scientific Analysis**

For any instument, the observed output from the system is convolution of the input with the instrument response. Here, in case of SoLEXS, the observed spectrum from the instrument is the incident spectrum convolved with the system transfer function. Thus, the knowledge of the system response function aids in spectral analysis through forward/reverse modelling. The two main components of the response are:
- Ancilliary Response Function (ARF) - it characterizies the effective area of the system
- Spectral Redistribution Function (SRF) - it is a probability distribution across the output energy bins and tells about the probability of incoming photon being registered in each bin.

For SDDs the SRF primarily consists of two Gaussian peaks - main and Si escape peak. For a monoenergetic beam of energy $E_0$, the main peak is centered on $E_0$ and a Si-escape peak occurs at an energy $E_0 - 1.74$ $keV$.

In the present exercise, we consider the system transfer functions and use the same to convolve with the model spectrum and use $\chi^2$ minimization algorithm to find the best fit between model and observed spectrum. From the best fit model spectrum, we estimate the properties of the source such as temperature, elemental abundance and alike.

```python
def timestamp2datetime(timestamp_arr):
    datetime_arr = []
    for ta in timestamp_arr:
        datetime_arr.append(datetime.datetime.fromtimestamp(ta))

    return datetime_arr

lc_datetime = timestamp2datetime(lc_data['TIME'])
```

```python
def rebin_lc(lc, time_arr ,rebin_sec): #lc: counts per second
    datetime_arr = time_arr
    extra_bins = len(lc) % rebin_sec
    if extra_bins != 0:
        lc = lc[:-extra_bins]
    new_bins = int(len(lc)/rebin_sec)
    new_lc = lc.reshape((new_bins, rebin_sec)).sum(axis=1)
    new_tm = np.arange(new_bins)*rebin_sec


    new_datetime_arr = []
    for ii in new_tm:
        new_datetime_arr.append(datetime_arr[int(ii)])

    new_lc = new_lc/rebin_sec 

    return new_lc, new_datetime_arr
```

```python
rebin_sec = 5
lc_60, time_arr_60 = rebin_lc(lc_data['COUNTS_ALL'],lc_data['TIME'],rebin_sec)
datetime_arr_60 = timestamp2datetime(time_arr_60)
```

### Model (Thermal Emission)

1. [Sunkit-Spex](https://github.com/sunpy/sunkit-spex)
2. [OSPEX](https://hesperia.gsfc.nasa.gov/~kim/rhessi3/software/spectroscopy/spectral-analysis-software/index.html)

```python
class f_vth_sherpa(model.ArithmeticModel):
    def __init__(self,name='f_vth'):
        self.temperature = model.Parameter(name,'Temperature',10,min=1,hard_min=0)
        self.em = model.Parameter(name,'EM',1,min=1e-5,hard_min=1e-49)
        model.ArithmeticModel.__init__(self, name,(self.temperature,self.em))

    def calc(self,pars,x,*args,**kwargs):
        inds = (x>1.0002920302956426) & (x<100)
        x = x[inds]
        x = np.append(x,2*x[-1]-x[-2])
        ene = x*u.keV
        temp = pars[0]*1e6*u.K
        em = pars[1]*1e49/(u.cm**5)

        y = thermal_emission(ene,temp,em,abundance_type='sun_coronal') # alternative: sun_photospheric

        spec = np.zeros(len(inds))
        spec[inds] = y

        return spec

mdl = f_vth_sherpa()
ene_x = pha.get_x()
print(mdl)
```

```python
fig, ax = plt.subplots()
ax.semilogy(ene_x,mdl.calc([5,1],ene_x),label='5 MK')
ax.semilogy(ene_x,mdl.calc([10,1],ene_x),label='10 MK')
ax.semilogy(ene_x,mdl.calc([20,1],ene_x),label='20 MK')
ax.semilogy(ene_x,mdl.calc([30,1],ene_x),label='30 MK')
ax.legend()
ax.set_xlabel('Energy (keV)')
ax.set_ylabel('Flux')
ax.set_xlim([1,20]);
ax.set_ylim([1,3e8]);
```

### Fitting to the Model

$$
\chi^2=\sum \frac{\left(C(I)-C_p(I)\right)^2}{(\sigma(I))^2}
$$


1. [XSPEC](https://heasarc.gsfc.nasa.gov/xanadu/xspec/)
2. [SPEX](https://www.sron.nl/astrophysics-spex/)
3. [ISIS](https://space.mit.edu/cxc/isis/)
4. [Sherpa](https://cxc.cfa.harvard.edu/sherpa/)

```python
pha = read_pha('spectral_data/AL1_SOLEXS_20240212_SDD2_L1_512_034615_034645.pha',use_errors=True)
pha.set_rmf(rmf)
pha.set_analysis('energy')
pha.notice(2.8,12)

pha.syserror = pha.counts*6e-2 # 6 percent

pha.staterror[pha.staterror==0] = 1
pha.syserror[pha.syserror == 0] = 1e-5

mdl_rsp.model.pars[0].val = 10 # Temperature
mdl_rsp.model.pars[1].val = 0.01 #EM

fi = Fit(pha,mdl_rsp)

fres = fi.fit()
print(fres)
print(f'Reduced Chi2 = {fres.statval/fres.dof:.4f}')
```
